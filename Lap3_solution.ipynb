Task 1 — Before Chart

إنشاء نسخة Before لاكتشاف مشاكل التصميم.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
   "Department": ["HR", "Finance", "IT", "Sales"],
   "Absence": [5, 8, 12, 7]
})

plt.bar(df["Department"], df["Absence"],
       color=["red", "green", "blue", "purple"])
plt.title("Employee Absence")
plt.xlabel("Department")
plt.ylabel("Absence %")
plt.grid(True)
plt.legend(["Absence"])
plt.show()
```

Task 2 — Palette

تطبيق Palette بسيطة بلون واحد.

```python
plt.figure(figsize=(8,5))
plt.bar(df["Department"], df["Absence"], color="steelblue")
plt.title("Employee Absence by Department")
plt.xlabel("Department")
plt.ylabel("Absence (%)")
plt.show()
```

Task 3 — Decluttering

إزالة العناصر البصرية الزائدة.

```python
plt.figure(figsize=(8,5))
plt.bar(df["Department"], df["Absence"], color="steelblue")
plt.title("Employee Absence by Department")
plt.xlabel("Department")
plt.ylabel("Absence (%)")
plt.grid(False)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.show()
```

Task 4 — Takeaway + Callout

إضافة Takeaway وCallout واحد.

```python
plt.figure(figsize=(8,5))
plt.bar(df["Department"], df["Absence"], color="steelblue")
plt.title("IT Has the Highest Absence — 12%")
plt.xlabel("Department")
plt.ylabel("Absence (%)")
plt.grid(False)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.annotate("Highest",
            xy=("IT", 12),
            xytext=("Finance", 13),
            arrowprops=dict(arrowstyle="->"))
plt.show()
```

Task 6 — Arabic / RTL

إنشاء نسخة عربية من الرسم.

```python
arabic = ["الموارد البشرية", "المالية", "تقنية المعلومات", "المبيعات"]

plt.figure(figsize=(9,5))
plt.bar(arabic, df["Absence"], color="steelblue")
plt.title("تقنية المعلومات لديها أعلى نسبة غياب — 12%")
plt.xlabel("القسم")
plt.ylabel("نسبة الغياب (%)")
plt.grid(False)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.show()
```
