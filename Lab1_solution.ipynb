الكود الأساسي

الاستخدام: يجهّز البيانات التي سنستخدمها في جميع الرسومات.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
   "Region": ["Riyadh", "Makkah", "Madinah", "Eastern", "Asir"],
   "Adoption": [85, 72, 91, 65, 78]
}

df = pd.DataFrame(data)

df
```


1. Sorted Bar

الاستخدام: يرتب المناطق حسب قيمة Adoption حتى تكون المقارنة أوضح.


```python
df_sorted = df.sort_values("Adoption")

plt.figure(figsize=(6,3))
plt.bar(df_sorted["Region"], df_sorted["Adoption"])
plt.xlabel("Region")
plt.ylabel("Adoption (%)")
plt.title("Digital Adoption by Region — Sorted")
plt.xticks(rotation=45)
plt.show()
```

2. Unsorted Bar

الاستخدام: يعرض نفس البيانات بالترتيب الأصلي بدون ترتيب حسب القيمة.

```python
plt.figure(figsize=(6,3))
plt.bar(df["Region"], df["Adoption"])
plt.xlabel("Region")
plt.ylabel("Adoption (%)")
plt.title("Digital Adoption by Region — Unsorted")
plt.xticks(rotation=45)
plt.show()
```

3. Pie

الاستخدام: يعرض البيانات على شكل مخطط دائري.

```python
plt.figure(figsize=(6,3))

plt.pie(
   df["Adoption"],
   labels=df["Region"],
   autopct="%1.1f%%"
)

plt.title("Digital Adoption by Region — Pie")
plt.show()
```

4. Bubble

الاستخدام: يمثل قيمة Adoption بحجم الفقاعة؛ كلما كبرت الفقاعة كانت القيمة أعلى.

```python
plt.figure(figsize=(6,3))

plt.scatter(
   df["Region"],
   df["Adoption"],
   s=df["Adoption"] * 20
)

plt.xlabel("Region")
plt.ylabel("Adoption (%)")
plt.title("Digital Adoption by Region — Bubble")
plt.xticks(rotation=45)
plt.show()

```

5. Single-row Heatmap

الاستخدام: يستخدم اللون لإظهار القيم الأعلى والأقل بسرعة.

```python
plt.figure(figsize=(5,1))

sns.heatmap(
   [df["Adoption"]],
   annot=True,
   xticklabels=df["Region"],
   yticklabels=["Adoption"],
   cmap="Blues"
)

plt.title("Digital Adoption by Region — Heatmap")
plt.show()
```

```python
df_sorted = df.sort_values(by="Adoption", ascending=True)

plt.figure(figsize=(5, 1))

sns.heatmap(
    [df_sorted["Adoption"].values],
    annot=True,
    xticklabels=df_sorted["Region"],
    yticklabels=["Adoption"],
    cmap="Blues"
)

plt.title("Digital Adoption by Region — Heatmap")
plt.show()
```
